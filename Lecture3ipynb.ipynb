```python
import numpy as np
```

```python
# 1D array
arr = np.array({10,20,30,40})
print(arr)
```

```python
# 2D array
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
```

```python
# 3D array
arr = np.array([[[1,2,3],[4,5,6]],[[5,6,7],[8,9,0]]])
print(arr)
```

```python
a=np.array([1,2,3,4])
b=np.zeros((3,4))
c=np.ones((2,3))
d=np.arange(0,20,2)
e=np.linspace(0,10,5)
print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
# Vectorized Computation
arr = np.array([1,2,3,4,5])
result = arr*2
print(result)
```

```python
list = (1,2,3,4,5)
result = list*2
print(result)
```

```python
arr = [1, 2, 3, 4, 5]

result = []

for i in arr:
    result.append(i * 2)

print(result)
```

```python
arr = np.array([1, 2, 3, 4, 5])
result = arr + 10
print(result)
```

```python
arr = np.array([1,4,9,16,18])
print(np.sqrt(arr))
```

```python
arr = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
print(np.array_equal(arr,arr2))
```

```python
arr = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
print(np.array(arr+arr2))
```
